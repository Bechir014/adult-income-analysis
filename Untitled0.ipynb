Importation des bibliothèques nécessaires

```python
# Importation des bibliothèques nécessaires

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import LabelEncoder, StandardScaler
```

Chargement du dataset Adult Census Income.

```python
# Définition des noms des colonnes

columns = [
    "age",
    "workclass",
    "fnlwgt",
    "education",
    "education-num",
    "marital-status",
    "occupation",
    "relationship",
    "race",
    "sex",
    "capital-gain",
    "capital-loss",
    "hours-per-week",
    "native-country",
    "income"
]
# Chargement du dataset

df = pd.read_csv(
    "adult.data",
    names=columns,
    sep=",",
    skipinitialspace=True
)
# Affichage des premières lignes du dataset

df.head()
```

```python
# Dimensions du dataset

df.shape
```

Exploration initiale des données (head, info, describe).

```python
# Affichage des 5 dernières lignes

df.tail()
```

```python
# Informations générales du dataset

df.info()
```

```python
# Statistiques des variables numériques

df.describe()
```

```python
# Affichage des colonnes

df.columns
```

Nettoyage des données (valeurs manquantes, doublons, préparation).

```python
# Vérification des valeurs inconnues avec ?

(df == "?").sum()
```

```python
# Remplacer les valeurs inconnues ? par NaN et regarde celle d avant tu vas trouver 0

df.replace("?", np.nan, inplace=True)
```

```python
# Nombre de valeurs manquantes par colonne

df.isnull().sum()
```

```python
# Suppression des lignes avec valeurs manquantes

df.dropna(inplace=True)
```

```python
# Vérification finale

df.isnull().sum()
```

```python
# Nombre de doublons

df.duplicated().sum()
```

```python
# Suppression des doublons

df.drop_duplicates(inplace=True)
```

```python
# Dimensions après nettoyage

df.shape
```

Transformation des données (Encodage des variables catégorielles)

```python
# Affichage des valeurs des variables principales

print(df["workclass"].unique())
print(df["occupation"].unique())
print(df["income"].unique())
```

```python
# Création de l'objet LabelEncoder

encoder = LabelEncoder()
# Encodage de la variable occupation

df["occupation"] = encoder.fit_transform(df["occupation"])

# Encodage de la variable income

df["income"] = encoder.fit_transform(df["income"])

# Affichage des premières lignes après transformation


# Encodage des variables catégorielles

encoder = LabelEncoder()

for col in df.select_dtypes(include="object").columns:
    df[col] = encoder.fit_transform(df[col])

df.head()
```

```python
# Vérifier les types des variables

df.info()
```

Normalisation des données avec StandardScaler

```python
# Sélection des colonnes numériques

numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

numeric_columns
```

```python
# Création du normalisateur

scaler = StandardScaler()

# Normalisation des variables numériques

df[numeric_columns] = scaler.fit_transform(df[numeric_columns])

 # Affichage après normalisation

df.head()
```

```python
# Vérification de la normalisation

df[numeric_columns].describe()
```

Matrice de corrélation et Heatmap.

La matrice de corrélation permet de mesurer la relation entre toutes les variables numériques du dataset.

```python
# Calcul de la matrice de corrélation

corr_matrix = df.corr()

corr_matrix

# Affichage de la matrice de corrélation

plt.figure(figsize=(12,8))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm"
)

plt.title("Matrice de corrélation des variables")

plt.show()
```

Sélection des variables importantes et création du dataset final

```python
# Sélection des variables finales

selected_columns = [
    "occupation",
    "workclass",
    "income",
    "education-num",
    "age",
    "hours-per-week",
    "capital-gain"
]

df_final = df[selected_columns]

# Afficher les premières lignes

df_final.head()

# Informations sur le dataset final

df_final.info()

# Nombre de lignes et colonnes

df_final.shape
```

Analyse finale et interprétation des résultats

```python
# Analyse du revenu selon workclass

workclass_income = pd.crosstab(
    df_final["workclass"],
    df_final["income"],
    normalize="index"
)

workclass_income
```

```python
plt.figure(figsize=(10,5))

workclass_income.plot(
    kind="bar",
    stacked=True,
    figsize=(10,5)
)

plt.title("Répartition des revenus selon le secteur d'activité")
plt.xlabel("Workclass")
plt.ylabel("Proportion")

plt.show()
```

```python
# Analyse du revenu selon occupation

occupation_income = pd.crosstab(
    df_final["occupation"],
    df_final["income"],
    normalize="index"
)

occupation_income
```

```python
plt.figure(figsize=(12,5))

occupation_income.plot(
    kind="bar",
    stacked=True,
    figsize=(12,5)
)

plt.title("Répartition des revenus selon la profession")
plt.xlabel("Occupation")
plt.ylabel("Proportion")

plt.show()
```
